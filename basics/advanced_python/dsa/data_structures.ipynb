### List
Lists are ordered, mutable, and can contain elements of different types. 

```python
# Create an algorithm to find the lowest value in a list
my_array = [7, 12, 9, 4, 11, 8]
minVal = my_array[0]
minVal = [i for i in my_array if i < minVal][0]
print(minVal)

# Since one value must be comapred with all elements of the list, 
# the time complexity of this algorithm is O(n), i.e Linear. 
```

### Stacks
Stack follows Last-In-First-Out (LIFO) principle. <br>
A Stack can hold values of multiple data-types. <br>
Implemented using arrays or linked-lists, as python doesn't have a dedicated <b>Stack</b> class. Implementing as a custom class gives better encapsulation. <br>

<b>Usage</b>
Stacks can be used to implement undo mechanisms, to revert to previous states, to create algorithms for depth-first search in graphs, or for backtracking.

<b>Methods</b>
- Push: Adds a new element on the stack.
- Pop: Removes and returns the top element from the stack.
- Peek: Returns the top (last) element on the stack.
- isEmpty: Checks if the stack is empty.
- Size: Finds the number of elements in the stack.

```python
stack = Stack()

stack.push(1)
print(stack.stack) 
stack.push(5)
print(stack.stack)
stack.push(9)
print(stack.stack)
print(stack.peek())
stack.pop()
print(stack.stack)
print(stack.isEmpty())
stack.pop()
print(stack.stack)
print(stack.peek())
stack.pop()
print(stack.peek())

    
```

### Queue
Queues follow First-in-First-Out (FIFO). <br>
Can be implemented using arrays or linked-lists. 

<b>Methods</b>
- Enqueue: Adds a new element to the queue.
- Dequeue: Removes and returns the first (front) element from the queue.
- Peek: Returns the first element in the queue.
- isEmpty: Checks if the queue is empty.
- Size: Finds the number of elements in the queue.

```python
# Create a queue
myQueue = Queue()

myQueue.enqueue('A')
myQueue.enqueue('B')
myQueue.enqueue('C')

print("Queue: ", myQueue.queue)
print("Peek: ", myQueue.peek())
print("Dequeue: ", myQueue.dequeue())
print("Queue after Dequeue: ", myQueue.queue)
print("isEmpty: ", myQueue.isEmpty())
print("Size: ", myQueue.size())
```

### LinkedList

```python
node1 = Node(7)
node2 = Node(11)
node3 = Node(3)
node4 = Node(2)
node5 = Node(9)
node6 = Node(10)

node1.next = node2
node2.next = node3
node3.next = node4
node4.next = node5

linkedList = LinkedList()
linkedList.head = node1
print("LL Before Deletion:") 
linkedList.traverseAndPrint()
print("Lowest Value in LL:")
linkedList.findLowestValue()
linkedList.deleteSpecificNode(node6)
print("LL After Deletion:")
linkedList.traverseAndPrint()

# Insert a new node with value 97 at position 2
newNode = Node(97)
try:
    linkedList.insertNodeAtPosition(newNode, 12)
except IndexError as error:
    print(error)
print("LL After Insertion:")
linkedList.traverseAndPrint()
```

### Hash Table

<b>Steps to build a hash table</b>:<br>

1) Create an empty list / dict / set
2) Create a hash function
3) Inserting an element using a hash function
4) Looking up an element using a hash function
5) Handling collisions


Hash Tables are great for:
- Checking if something is in a collection (like finding a book in a library).
- Storing unique items and quickly finding them (like storing phone numbers).
- Connecting values to keys (like linking names to phone numbers).

Arrays and Linked Lists have time complexity O(n) for search and delete, while Hash Tables have just O(1) on average.

<b>Summary</b>
- Hash Table elements are stored in storage containers called buckets.
- A hash function takes the key of an element to generate a hash code.
- The hash code says what bucket the element belongs to, so now we can go directly to that Hash Table element: to modify it, or to delete it, or just to check if it exists.
- A collision happens when two Hash Table elements have the same hash code, because that means they belong to the same bucket.
- Collision can be solved by Chaining by using lists to allow more than one element in the same bucket.

```python
# 1) create an empty list
my_list = []
for i in range(10): # 10 empty elements
    my_list.append(None) 
# print(my_list)
# each of these elements is called a bucked in a hash table

# 2) create a hash function - upto the creator of the hash table
# common way - find a way to convert the value into a number,
# that equals one of the hash table's index

# In this example, we use the Unicode number of each character, 
# summarize them and do a modulo 10 operation to get index numbers 0-9
def hash_function(value):
    sum_of_chars = 0
    for char in value:
        sum_of_chars += ord(char)
    return sum_of_chars % 10
print("Bob has a hash code", hash_function("Bob"))
# The character B has Unicode number 66, o has 111, and b has 98. 
# Adding those together we get 275. Modulo 10 of 275 is 5, so "Bob" should be stored at index 5.

# The number returned by the hash function is called the "hash code".

# 3) Inserting an Element
# According to our hash function, "Bob" should be stored at Index 5
# Now we create a function that adds items to our hash table
def add(name): # "add" is not a built-in function in python
    index = hash_function(name)
    my_list[index] = name
add("Bob")
add("Pete")
add("John")
add("Lisa")
print(my_list)
# Note that names with same hash values will overwrite one another in this case

# 4) Looking up a name
def contains(name):
    index = hash_function(name)
    return my_list[index] == name
print(f"Pete is in the hash table? {contains("Pete")}")
# now  we do not have to check element by element to find out if "Pete" is in there,
# we can just use the hash function to go straight to the right element!

# 5) Handling collisions
# Adding "Stuart" will cause a collision with "Lisa" as the computed unicode hash is 3 (same)
# "Chaining" is a solution, start by creating another empty list in the following fashion
my_list = []
for i in range(10):
    my_list.append([])
print(my_list)

# rewriting add function to add the same names as before
def add(name):
    index = hash_function(name)
    my_list[index].append(name)
add("Bob")
add("Pete")
add("John")
add("Lisa")
add("Stuart")
print(my_list) # The "Lisa"-"Stuart" problem is now sorted
# After implementing each bucket as a list, "Stuart" can also be stored at index 3

# re-write the contains function to be compatible with the listed bucket type
def contains(name):
    index = hash_function(name)
    return name in my_list[index]
print(f"Stuart is in the hash table? {contains("Stuart")}")
# Searching for "Stuart" now takes a little bit longer time,
# because we also find "Lisa" in the same bucket,
# but still much faster than searching the entire Hash Table.
```
