### List
Lists are ordered, mutable, and can contain elements of different types. 

```python
# Create an algorithm to find the lowest value in a list
my_array = [7, 12, 9, 4, 11, 8]
minVal = my_array[0]
minVal = [i for i in my_array if i < minVal][0]
print(minVal)

# Since one value must be comapred with all elements of the list, 
# the time complexity of this algorithm is O(n), i.e Linear. 
```

### Stacks
Stack follows Last-In-First-Out (LIFO) principle. <br>
A Stack can hold values of multiple data-types. <br>
Implemented using arrays or linked-lists, as python doesn't have a dedicated <b>Stack</b> class. Implementing as a custom class gives better encapsulation. <br>

<b>Usage</b>
Stacks can be used to implement undo mechanisms, to revert to previous states, to create algorithms for depth-first search in graphs, or for backtracking.

<b>Methods</b>
- Push: Adds a new element on the stack.
- Pop: Removes and returns the top element from the stack.
- Peek: Returns the top (last) element on the stack.
- isEmpty: Checks if the stack is empty.
- Size: Finds the number of elements in the stack.

```python
from dsa_utils import Stack
stack = Stack()

stack.push(1)
print(stack.stack) 
stack.push(5)
print(stack.stack)
stack.push(9)
print(stack.stack)
print(stack.peek())
stack.pop()
print(stack.stack)
print(stack.isEmpty())
stack.pop()
print(stack.stack)
print(stack.peek())
stack.pop()
print(stack.peek())

    
```
